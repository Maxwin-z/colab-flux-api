# FLUX Image Generator · Colab 启动

**一次性配置(每个 Colab 账号只做一次):** 左侧 🔑 **Secrets** 面板里添加两个密钥,并各自打开 **Notebook access**:

| 名称 | 值 |
|---|---|
| `HF_TOKEN` | HuggingFace token(下载 FLUX 权重用) |
| `CLOUDFLARE_TUNNEL_TOKEN` | Cloudflare 隧道 token(固定域名用;不配则自动退回随机 trycloudflare 地址) |

之后每次只需运行下面这一格。加载完模型后日志会打印 `publishing on https://flux.codecrab.dev`,直接访问即可(固定地址、重启不变、掉线自动重连)。


```python
import os
from google.colab import userdata

# --- 密钥(在 🔑 面板里配好,开 Notebook access) ---
os.environ['HF_TOKEN'] = userdata.get('HF_TOKEN')                 # HuggingFace,下载 FLUX 权重
try:
    os.environ['CLOUDFLARE_TUNNEL_TOKEN'] = userdata.get('CLOUDFLARE_TUNNEL_TOKEN')  # 固定域名隧道
except Exception as e:
    print(f'[colab] 未取到 CLOUDFLARE_TUNNEL_TOKEN ({e}),将退回随机 trycloudflare 地址')
# TUNNEL_HOSTNAME 默认 flux.codecrab.dev(见 run_colab.py);要换域名再取消下一行注释:
# os.environ['TUNNEL_HOSTNAME'] = 'flux.codecrab.dev'

# --- 拉取 / 更新代码 ---
repo_dir = '/content/colab-flux-api'
if os.path.exists(repo_dir):
    os.chdir(repo_dir)
    !git pull
else:
    !git clone https://github.com/Maxwin-z/colab-flux-api.git {repo_dir}
    os.chdir(repo_dir)

# --- 装依赖并启动(发布在 https://flux.codecrab.dev) ---
!pip install -r requirements.txt -q
!python run_colab.py
```
